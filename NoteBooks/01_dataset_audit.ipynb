```python
import json
import pandas as pd


def parse_json(value):
    if value is None or value == "":
        return None

    try:
        return json.loads(value)
    except (json.JSONDecodeError, TypeError):
        return None


train_df["tools_parsed"] = train_df["tools"].apply(parse_json)
test_df["tools_parsed"] = test_df["tools"].apply(parse_json)
```

```python
print("Train rows:", len(train_df))
print("Test rows:", len(test_df))

print(
    "Train parsing failures:",
    train_df["tools_parsed"].isna().sum()
)

print(
    "Test parsing failures:",
    test_df["tools_parsed"].isna().sum()
)
```

```python
print(
    train_df["tools_parsed"]
    .apply(lambda x: isinstance(x, list))
    .value_counts()
)
```

```python
def has_tools(x):
    return isinstance(x, list) and len(x) > 0


train_df["has_tools"] = train_df["tools_parsed"].apply(has_tools)
test_df["has_tools"] = test_df["tools_parsed"].apply(has_tools)

print("TRAIN")
print(train_df["has_tools"].value_counts())
print(train_df["has_tools"].value_counts(normalize=True) * 100)

print("\nTEST")
print(test_df["has_tools"].value_counts())
print(test_df["has_tools"].value_counts(normalize=True) * 100)
```

```python
from collections import Counter


def extract_tool_names(tools):
    if not isinstance(tools, list):
        return []

    names = []

    for tool in tools:
        if not isinstance(tool, dict):
            continue

        function = tool.get("function", {})
        name = function.get("name")

        if name:
            names.append(name)

    return names


train_df["tool_names"] = train_df["tools_parsed"].apply(extract_tool_names)
test_df["tool_names"] = test_df["tools_parsed"].apply(extract_tool_names)


train_tool_counter = Counter(
    name
    for names in train_df["tool_names"]
    for name in names
)

test_tool_counter = Counter(
    name
    for names in test_df["tool_names"]
    for name in names
)

print("Unique tools - train:", len(train_tool_counter))
print("Unique tools - test:", len(test_tool_counter))

print("\nTRAIN")
print(train_tool_counter.most_common())

print("\nTEST")
print(test_tool_counter.most_common())
```

```python
tool_schemas = {}

for _, row in train_df[train_df["has_tools"]].iterrows():
    for tool in row["tools_parsed"]:
        function = tool.get("function", {})
        name = function.get("name")

        if name and name not in tool_schemas:
            tool_schemas[name] = function


for name, schema in tool_schemas.items():
    parameters = schema.get("parameters", {})

    print("=" * 80)
    print(name)
    print("Description:", schema.get("description"))
    print("Properties:", list(parameters.get("properties", {}).keys()))
    print("Required:", parameters.get("required", []))
```

```python
import json


def extract_tool_call(response):
    if not isinstance(response, str):
        return None

    decoder = json.JSONDecoder()

    # Try every possible JSON-object starting position
    for i, char in enumerate(response):
        if char != "{":
            continue

        try:
            obj, _ = decoder.raw_decode(response[i:])

            if (
                isinstance(obj, dict)
                and "name" in obj
                and "arguments" in obj
            ):
                return obj

        except json.JSONDecodeError:
            continue

    return None


tool_response_df["tool_call"] = (
    tool_response_df["response"]
    .apply(extract_tool_call)
)

print(
    "Successfully parsed:",
    tool_response_df["tool_call"].notna().sum()
)

print(
    "Failed to parse:",
    tool_response_df["tool_call"].isna().sum()
)
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

MODEL_NAME = "Qwen/Qwen2.5-7B-Instruct"

quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=quantization_config,
    device_map="auto",
)

model.eval()

print("Model loaded successfully!")
```

```python
sample = no_detected_tool.iloc[0]

prompt = f"""
You are auditing an e-commerce customer support dataset.

Analyze the assistant's actual behavior based ONLY on the user prompt
and the reference response.

User prompt:
{sample["prompt"]}

Reference response:
{sample["response"]}

Classify the behavior into exactly one of these labels:

TOOL_CALL
DIRECT_ANSWER
CLARIFICATION
HUMAN_ESCALATION
MIXED
OTHER

Also identify whether the assistant actually called a tool.

Return ONLY valid JSON in this format:

{{
  "behavior": "...",
  "tool_called": true,
  "tool_name": "..."
}}

If no tool was called, use:
"tool_called": false
"tool_name": null
"""

messages = [
    {"role": "system", "content": "You are a precise dataset auditor. Return JSON only."},
    {"role": "user", "content": prompt}
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(text, return_tensors="pt").to(model.device)

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=150,
        temperature=0.0,
        do_sample=False,
    )

result = tokenizer.decode(
    outputs[0][inputs["input_ids"].shape[1]:],
    skip_special_tokens=True
)

print("=" * 100)
print(f"result \n {result}")
print("=" * 100)
print(f"prompt \n {sample["prompt"]}")
print("=" * 100)
print(f"response: \n {sample["response"]}")
```

```python
def detecte_semantic(sample):
    prompt = f"""
    You are auditing an e-commerce customer support dataset.

    Analyze the assistant's actual behavior based ONLY on the user prompt
    and the reference response.

    User prompt:
    {sample["prompt"]}

    Reference response:
    {sample["response"]}

    Classify the behavior into exactly one of these labels:

    TOOL_CALL
    DIRECT_ANSWER
    CLARIFICATION
    HUMAN_ESCALATION
    MIXED
    OTHER

    Also identify whether the assistant actually called a tool.

    Return ONLY valid JSON in this format:

    {{
      "behavior": "...",
      "tool_called": true,
      "tool_name": "..."
    }}

    If no tool was called, use:
    "tool_called": false
    "tool_name": null
    """

    messages = [
        {"role": "system", "content": "You are a precise dataset auditor. Return JSON only."},
        {"role": "user", "content": prompt}
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(text, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=150,
            temperature=0.0,
            do_sample=False,
        )

    result = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return result
```

```python
import json

results = []
sampled = no_detected_tool.sample(n=100, random_state=42)

for i, (_, sample) in enumerate(sampled.iterrows(), start=1):

    raw_result_str = detecte_semantic(sample)
    parsed_result_dict = json.loads(raw_result_str)
    results.append({
        "prompt": sample["prompt"],
        "response": sample["response"],
        **parsed_result_dict
    })

    if i % 10 == 0:
        print(f"Processed {i}/100 samples")
```
